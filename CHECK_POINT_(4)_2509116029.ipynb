### **Outliers Values**

Pada tahap ini dilakukan identifikasi nilai ekstrem (outliers) pada kolom yang bertipe data numerik, seperti RAM, Storage, dan Final Price. Pemeriksaan ini bertujuan untuk insight yang akurat sehingga dapat memperkuat analisis dan model yang akan kita bangun nantinya.


Akan tetapi,  dalam industri smartphone, variasi spesifikasi dan harga yang cukup besar merupakan hal yang wajar, terutama karena adanya perbedaan segmen pasar antar level sehingga tidak perlu ditangani.

Berdasarkan hasil identifikasi sebagai berikut:

* RAM: 0%
* Storage: sekitar 0,9%
* Final Price: sekitar 5,1% (kurang dari 1%)

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

# **Construct Data**

**1. Membuat Kolom Baru "Kategori Harga"**

Kolom ini diperlukan untuk mengelompokan harga untuk membagi smartphone ke dalam beberapa segmen pasar, sehingga pola hubungan antara RAM, storage, dan brand terhadap harga dapat terlihat lebih jelas.

Dengan adanya kategori harga, analisis menjadi lebih terstruktur dan mudah dibandingkan.

```python
def kategori_harga(harga):
    if harga < 200 and harga >=100:
        return 'Entry Level'
    elif harga < 400:
        return 'Mid Range'
    elif harga < 700:
        return 'Upper Mid'
    elif harga < 1000:
        return 'Premium'
    elif harga >= 1000 and harga <= 10000:
        return 'High End'
    else:
        return 'berhenti'

df['Kategori Harga'] = df['Final Price'].apply(kategori_harga)
```

# **Bubble Chart**

```python
plt.figure(figsize=(14, 8))

# Buat ukuran bubble berdasarkan harga (lebih besar = lebih mahal)
scatter = sns.scatterplot(
    x='RAM',
    y='Storage',
    size='Final Price',
    hue='Kategori Harga',
    data=df,
    sizes=(20, 1000),        # Ukuran bubble proporsional dengan harga
    alpha=0.6,
    palette='Spectral',
    edgecolor='black',
    linewidth=0.5
)

plt.title('Posisi Produk Berdasarkan RAM dan Storage (Ukuran: Harga)', fontsize=16, fontweight='bold')
plt.xlabel('RAM (GB)', fontsize=12)
plt.ylabel('Storage (GB)', fontsize=12)

# Atur sumbu x (RAM biasanya 2,3,4,6,8,12,16)
plt.xticks([2,3,4,6,8,12,16])

plt.grid(True, linestyle='--', alpha=0.3)
plt.legend(title='Kategori Harga', bbox_to_anchor=(1.15, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

Visualisasi ini menunjukkan hubungan antara RAM, storage, dan harga smartphone. Dari grafik terlihat bahwa smartphone dengan RAM dan storage yang lebih besar cenderung memiliki harga yang lebih tinggi, sedangkan smartphone dengan spesifikasi lebih rendah umumnya berada pada rentang harga yang lebih rendah.

Tindakan yang bisa dilanjutkan yaitu dengan mengelompokkan smartphone berdasarkan kombinasi RAM, storage, dan harga untuk melihat pola produk-produk yang memiliki spesifikasi tinggi dengan harga kompetitif. Dengan tindakanini nantinya dapa mengidentifikasi produk-produk yang masuk kategori spesifikasi tinggi dengan harga relatif terjangkau serta produk premium dengan spesifikasi tertinggi. Informasi ini berguna untuk memahami posisi kompetitif setiap brand di pasar.
